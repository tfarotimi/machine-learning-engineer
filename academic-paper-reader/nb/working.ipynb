### Load PDFs and Extract Contents

```python
import pypdf as pdf
from pathlib import Path

#set path for folder containing pdfs 
pdf_path = Path(r'C:\\DS_ML\\machine-learning-engineer\\academic-paper-reader\\data\\')

#grab all files ending in ".pdf"from folder
files = pdf_path.glob("*.pdf")

papers_dict = {}

#for each paper in the file, create a pdf reader object and get each page,
# for each page, extract text and add it to a string containing text from previous pages
# store contents for each paper in a dictionary
for i, paper in enumerate(files):
    reader = pdf.PdfReader(paper)
    pages = reader.pages
     
    paper_name = paper.stem
    paper_contents = ""

    for j, page in enumerate(pages):
        page_content = page.extract_text()
        paper_contents += page_content

    papers_dict[paper_name] = paper_contents

    
```

```python
def clean(text):

    #remove references
    ref_start = text.rfind("References")

    if ref_start != -1:
        text = text[:ref_start]


    #remove hyphens and newlines
    text = text.replace("-\n", "")
    text = text.replace("\n", " ")    # <- this step went missing

    #replace unicode characters with letters
    text = text.replace("ﬁ", "fi").replace("ﬂ", "fl")

    return text

def chunk(text, n_chunk):
    chunk_dict = {}
    words = text.split()
    if len(words) < n_chunk:
        print("text is too short to chunk")
    else: 
        # counter = 0, use enumerate instead of manual counter
        for counter, i in enumerate(range(0, len(words), n_chunk)):
            chunk = words[i:i + n_chunk + 10] #each chunk includes ten extra words that are the first ten words of the next chunk, so that those 10 words are associated with both the words that precede them and those that follow them  
            chunk_dict[counter] = " ".join(chunk)

    
    return chunk_dict
        
        
```

```python
chunked = {}

for i in papers_dict:
    cleaned = clean(papers_dict[i])
    chunked[i] = chunk(cleaned, 200)
```
